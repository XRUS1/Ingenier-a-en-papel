# Ingeniería en Papel con Blender para Mecanismos de 180°

La ingeniería en papel (*paper engineering*) aplicada en Blender permite **diseñar, prototipar y simular mecanismos tridimensionales plegables** (como libros *pop-up*, tarjetas desplegables u origami) antes de cortarlos físicamente. 

Para crear pop-ups interactivos que respondan al movimiento de "abrir y cerrar un libro a 180 grados", los flujos de trabajo avanzados reemplazan la animación tradicional por sistemas de esqueletos (*armatures*), restricciones matemáticas y simulaciones físicas. Esto permite evaluar dobleces reales, colisiones y el apilamiento de las capas de papel sin deformar el material de manera irreal.


## 📐 Mecanismos Clave de 180° a Modelar

A diferencia de los mecanismos de 90 grados (que dependen de las tapas en ángulo), los de 180 grados utilizan la fuerza del movimiento horizontal para levantar las figuras cuando el libro se abre totalmente plano. Al modelar, debes estructurar geométricamente tres tipologías específicas extraídas de la literatura de ingeniería de papel:

*   **Pliegue en V (V-Fold) - Tipo Asimétrico:** Es la base de casi todo pop-up. Es una tira de papel doblada por la mitad cuyas patas se pegan a ambos lados del pliegue central del lomo. El ángulo del pliegue determina qué tan alto y hacia dónde se levanta el objeto.
*   **Planos Paralelos (Parallel Layers) con Pestañas Opuestas:** Capas de papel que permanecen paralelas a las páginas del libro. Utilizan pestañas de soporte que se doblan en sentido contrario para empujar la figura hacia arriba al abrirse.
*   **Mecanismos de Tracción (Pull-Tabs y Brazos articulados en Pivote):** Tiras ocultas que, al abrirse la página principal, tiran de otra sección para rotarla o pivotarla de forma asimétrica.


## 💻 Flujo de Trabajo Técnico en Blender

Para diseñar estos libros con éxito, el flujo de trabajo geométrico debe estructurarse de la siguiente manera:

`[Dibujo 2D (Grease Pencil)] ➔ [Estructura de Huesos (Armatures)] ➔ [Restricciones (Constraints)] ➔ [Despliegue (Paper Model)]`

1.  **Rigging con Armatures (El esqueleto):** Cada pliegue de papel real debe estar representado por un **hueso (*Bone*)**. Coloca el hueso principal justo en el eje del lomo del libro. Usa una cadena de huesos hijos para las pestañas y los pliegues en V.
2.  **Restricciones de Transformación (*Constraints*):** Para evitar que el papel se "estire" (lo cual es físicamente imposible), usa la restricción **Limit Distance** en los extremos de tus planos de papel. Esto simula la rigidez del material y te avisará si el mecanismo digital se rompería al abrirse.
3.  **Animación Técnica:** Usa **Transformation Constraints** o **Inverse Kinematics (IK)** para que el movimiento de la página izquierda (que rota de 0 a 180 grados) arrastre matemáticamente y de forma exacta a los huesos del mecanismo interno.


## 🔗 Referencias Técnicas en Blender y Add-ons

A continuación se detallan los recursos de soporte, manuales técnicos y extensiones indispensables con sus rutas y secciones exactas:

*   **Tutorial en Video - Simulación de Gravedad Inversa:** El video técnico [The BASICS of creating realistic pop ups in Blender (Minuto 1:15)](https://www.youtube.com/watch?v=0nN4EKwnx8M) describe específicamente cómo ir a `Scene Properties > Rigid Body World > Field Weights` y cambiar la gravedad a un valor negativo para simular el cierre invertido del papel sin interferencia manual.
*   **Foro Técnico - Solución de Rigging con 2 Armatures:** El hilo de soporte [Blender Stack Exchange - Rigging for Popup Book (Respuesta 1 por Duarte Farrajota)](https://blender.stackexchange.com/questions/207766/rigging-for-popup-book) detalla la configuración exacta de dos esqueletos: un *Armature* de control (huesos amarillos `Page_L` y `Page_R`) y un *Armature* subordinado (huesos naranjas `Popup_Center`, `Popup_L`, `Popup_R`) conectados vía restricciones *IK Constraints* dirigidas a los extremos para evitar el estiramiento de las caras.
*   **Manual de Usuario de Blender - Sección Add-ons:** La sección técnica del [Blender 4.1 Manual - Paper Model Add-on](https://docs.blender.org/manual/es/4.1/addons/import_export/paper_model.html)

detalla el uso del panel *Paper Model Islands* de la pestaña de opciones de Malla. Explica el funcionamiento de la herramienta *Unfold* para la asignación de prioridades de corte en aristas y la ejecución del operador *Create Numbers* para etiquetar solapas de pegado de forma correlativa.
*   **Repositorio de Automatización de Pliegues (Origamify):** El código fuente en [GitHub - aconz2/blender-addon-origamify](https://github.com/aconz2/blender-addon-origamify) añade el operador ejecutable en la ruta de menú `Object > Origamify` empleando el script `origami.py`. Este calcula matemáticamente los patrones basados en las fórmulas de simulación de origami rígido descritas por Tomohiro Tachi. Su aplicación práctica sobre mallas se observa en
https://origamisimulator.org/ <br>





## 📚 Referencias Teóricas de Ingeniería en Papel (Mecanismos de 180°)

Dado que Blender es solo un entorno de representación geométrica, necesitas aplicar las reglas físicas y matemáticas del papel real recopiladas en las siguientes fuentes exactas:

*   **Masterclass en Video - Geometría de Pliegues en V:** La lista de reproducción [The Pop-Up Channel - Masterclass (Lecciones 3, 4 y 5)](https://youtube.com) de Duncan Birmingham explica las reglas matemáticas de los ángulos agudos y los planos de elevación asimétricos, detallando por qué los puntos de anclaje deben coincidir de forma simétrica respecto a la bisectriz del lomo para abrirse planos a 180°.
*   **Libro de Texto de Referencia - Capítulos Técnicos:** *Pop-Up Design and Paper Mechanics* de Duncan Birmingham. Debes consultar específicamente el **Capítulo 5: "The V-Fold Mechanism"** y el **Capítulo 8: "Generation of Parallel Layers"**, los cuales contienen los teoremas de geometría plana necesarios para calcular la longitud máxima de las pestañas de soporte antes de que colisionen en el centro del lomo.
*   **Demostración Práctica de Colapso (Modelos Físicos):** El video de análisis [The Elements of Pop-Up - Mecanismo de la Figura 24](https://youtube.com) (basado en el libro interactivo de David A. Carter y James Diaz) ilustra visualmente el comportamiento cinemático real de un cubo tridimensional de 180° que colapsa totalmente plano a 0° empleando bisagras asimétricas duales.
