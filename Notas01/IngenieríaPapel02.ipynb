# 📄 Ingeniería en Papel: Referencias y Obra de Yoojin Kim

**Yoojin Kim** es una destacada diseñadora e ingeniera de papel reconocida a nivel internacional. Su trabajo combina el arte botánico con mecanismos tridimensionales complejos. Actualmente, se desempeña como Directora de Arte de la marca de papelería de lujo **UWP Luxe** (una división de *Up With Paper*) y colabora de forma continua con la editorial *Jumping Jack Press*.

---

## 📚 1. Libros Pop-Up Publicados

A través de la ingeniería de papel, Yoojin ha logrado capturar la esencia de la naturaleza y las estaciones del año en formato editorial:

*   **Leaves: An Autumn Pop-Up Book (2017)**
    *   *Colaboradores:* Texto de Janet Lawler e ilustraciones de Lindsay Dale-Scott.
    *   *Descripción:* Su primer libro publicado. Explora la flora y fauna del otoño mediante el despliegue tridimensional de hojas en movimiento, hongos y animales del bosque. Ha sido traducido al francés y al japonés.
*   **Shells: A Pop-Up Book of Wonder (2019)**
    *   *Colaboradores:* Escrito por Janet Lawler.
    *   *Descripción:* Se enfoca en recrear las complejas formas geométricas y orgánicas de los caracoles y conchas marinas que se encuentran en las costas.
*   **Flora: A Botanical Pop-Up Book (2020)**
    *   *Colaboradores:* Texto de Nicole Yen e ilustraciones de Kathryn Selbert.
    *   *Descripción:* Una pieza dedicada a la primavera que recrea intrincados arreglos botánicos y flores que se abren de manera realista al pasar la página.
*   **Snowscape: A Winter Pop-Up Book (2022)**
    *   *Descripción:* Cierra el ciclo de las estaciones explorando paisajes invernales, animales en hibernación y nieve acumulada mediante pestañas de arrastre (*pull-tabs*) y capas interactivas.

## 🏆 2. Proyectos Comerciales y Premios

En su faceta de diseño industrial y papelería fina para **UWP Luxe**, ha empujado los límites de las tarjetas de felicitación interactivas:

*   **Línea de Tarjetas Estacionales:** Creadora de colecciones tridimensionales icónicas como *Winter Wonderland*, *Magnolia*, *Sugar Skull* (Día de Muertos) y *Cuckoo Clock*.
*   **Premio Louie Award:** Ganadora de este prestigioso galardón de la industria de la papelería en la categoría de tarjetas navideñas gracias a su diseño *"Eat, Drink, and Be Merry"*, reconocido por su nivel de detalle mecánico.
